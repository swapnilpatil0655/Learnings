### Operator Overloading

Operator overloading allows you to define the behavior of operators (+,-,*,etc.) for custom objects. You achieve this by overriding specific magic methods in your class.


`Common operator Overloading magic methods `
- `__add__(self, other)`: Adds two objects using the + operator.
- `__sub__(self, other)`: Subtracts two objects using the - operator.
- `__mul__(self, other)`: Multiplies two objects using * operator.
- `__truediv__(self, other)`:Divides two objects using the / operator.
- `__eq__(self, other)`: Check if two objects are equal using the == operator.
- `__lt__(self, other)`: Check if one object is less that another using the < operator>

<span style="color:#16a34a; font-weight:700; font-size:20px;">Common Operator Overloading Magic Methods</span>

- <span style="color:#ef4444; font-weight:700;">\_\_add\_\_(self, other)</span>: Adds two objects using the <code>+</code> operator.  
- <span style="color:#3b82f6; font-weight:700;">\_\_sub\_\_(self, other)</span>: Subtracts two objects using the <code>-</code> operator.  
- <span style="color:#f59e0b; font-weight:700;">\_\_mul\_\_(self, other)</span>: Multiplies two objects using the <code>*</code> operator.  
- <span style="color:#14b8a6; font-weight:700;">\_\_truediv\_\_(self, other)</span>: Divides two objects using the <code>/</code> operator.  
- <span style="color:#a855f7; font-weight:700;">\_\_eq\_\_(self, other)</span>: Checks if two objects are equal using the <code>==</code> operator.  
- <span style="color:#ec4899; font-weight:700;">\_\_lt\_\_(self, other)</span>: Checks if one object is less than another using the <code>&lt;</code> operator.
