This lesson covers the basics of creating classes and objects, including instance variables and methods.

```python
### Instance variable and methods

class Dog:
    ## Constructor
    def __init__(self,name,age):   ## self is used to refer to the current object instance inside a class method.
        self.name = name            ## # stores data in this object and we call it instance variable.
        self.age = age


## Crete objects

dog1=Dog("Buddy",3)
print(dog1)        ## This shows the location 

print(dog1.name)    ##  # accesses this object's data
print(dog1.age)
```

```python
dog2 = Dog("Lucy",2)
print(dog2.name)
print(dog2.age)
```

```python
## Define a class with instance methods
class Dog:
    def __init__(self,name,age):   
        ## Creating a constructor
        self.name = name
        self.age = age

    ## Creating a method
    def bark(self):
        print(f"{self.name} says woof")

dog1=Dog("Buddy",3)
dog1.bark()

dog2=Dog("Lucy",2)
dog2.bark()
```

```python
### Modeling a Bank Account

## Define a class for bank account
class BankAccount:
    def __init__(self,owner,balance=0):
        self.owner = owner       ## instance variable 
        self.balance = balance

    ## instance methods
    def deposite(self,amount):

        self.balance += amount
        print(f"{amount} is deposite to your bank account")
        print(f"Available balance is : {self.balance}")

    def withdrown(self,amount):

        if self.balance < amount:
            print("No sufficient balance")

        else:
            self.balance -= amount
            print(f"{amount} is withdrown from {self.owner}'s account")
            print(f"Available balance is : {self.balance}")

    
    def get_balance(self):
        return self.balance

    def check_balance(self):
        print(f"Your current balance is : {self.balance}")

## Create an account(object)

account =BankAccount("Swapnil",20000)

print(account.balance)
```

```python
## Call instance method

account.deposite(500)
```

```python
account.withdrown(500)
```

```python
account.withdrown(50000)

account.check_balance()
```

### Instance Variable
- `self.owner` and `self.balance` are **instance variables**.
- They store data that belongs to each bank account object.

### Instance Method
- `deposite()`, `withdrown()`, `get_balance()`, and `check_balance()` are **instance methods**.
- They use `self` to access and modify instance variables.

### Relationship in Your Program
- Instance variables store the account data.
- Instance methods perform actions on that data.
- Example:
  - `deposite()` increases  `self.balance`
  - `withdrown()` decreases  `self.balance`
  - `check_balance()` reads `self.balance`
