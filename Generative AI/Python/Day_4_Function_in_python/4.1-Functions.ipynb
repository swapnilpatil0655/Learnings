### 5. Default parameters

```python
## Default parameters

def greet(name):
    print(f"Hello {name}, Good Morning!")

# calling the function
greet("Swapnil")
```

```python
def greet(name):
    print(f"Hello {name}, Good Morning!")

greet()
```

```python
def greet(name = "Guest"):
    print(f"Hello {name}, Good Morning!")

greet()
```

### 6. Variable Length Arguments

#### Positional And Keyworkds arguments


```python
# Positional Arguments
def print_numbers(*swapnil): # *args
    for numbers in swapnil:
        print(numbers)
```

```python
print_numbers(1,2,3,4,5,6,7,8,"Swapnil") #*args allows us to pass a variable number of arguments to a function. It collects all the positional arguments into a tuple, which can be accessed within the function. In this example, we can pass any number of arguments to the print_numbers function, and it will print each of them on a new line.
```

```python
## Keyword Arguments

def print_details(**kwargs): # in keyword arguments all the key will be in the form of key value pair and it will be in the form of dictionary
    for key, value in kwargs.items():
        print(f"{key}: {value}")
```

```python
print_details(name="swapnil", age=30, city="Nagpur")
```

```python
def print_details(*args,**kwargs): 
    for val in args:
        print(f"Postional argument: {val}")

    for key, value in kwargs.items():
        print(f"Keyword argument: {key}: {value}")
```

```python
print_details(1,2,3,4,5,"Swapnil",name="swapnil", age=30, city="Nagpur")
```
