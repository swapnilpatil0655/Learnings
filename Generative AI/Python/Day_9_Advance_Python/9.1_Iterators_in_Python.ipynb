### Iterators

Iterators are advance Python concepts that allow for efficient looping and memory management. Iterators provide a way to access elements of a collection sequentially without exposing the underlying structure.

```python
my_list=[1,2,3,4,5,6]
for i in my_list:
    print(i)
```

```python
type(my_list)
```

```python
print(my_list)
```

## Iterator

```python
iterator = iter(my_list)
print(type(iterator))
```

```python
iterator
```

### Iterate through all the elements

```python
next(iterator) # every time you iterate net command will give you next iterator in the list and in the `StopIteration` error.
```

```python
iterator = iter(my_list)
```

```python
try:
    print(next(iterator))

except StopIteration:
    print("There are no element in the iterator")
```

```python
## String iterator

my_string = "Hello"
string_iterator = iter(my_string)

print(next(string_iterator))  #Output: H
print(next(string_iterator))  #Output: e
```
