### Assignment 1: Handling Division by zero

Write a function that two integer as input and return their division. Use try, except, and finally blocks to handle division by zero and print and appropriate message.

```python
a = int(input("Enter the 1st number:  "))
b = int(input("Enter the 2nd number:  "))
c =0
try:
    c = a/b
    print(f"Result is : {c}")

except ZeroDivisionError as ze:
    print(f"value of 2nd number is: {b}")
    print("Please enter value greater than Zero(0)")

finally:
    print("Execution is completed")
```

```python
def divide(a,b):
    try:
        result = a/b
        # print(f"Result:{result}")
        
    except ZeroDivisionError as ze:
        print(f"Error : {ze}")
        result = None
    finally:
        print("Execution completed.")

    return result

print(divide(2,2))
```

### Assignment 2: File Reading with exception handling

Write a function that reads the contents of a file name `data.txt`. Use try, except and finally blocks to handle file not found errors and ensure the file is properly closed.

```python
def readfile(filename):
    try:
        data = open (filename,'r')
        content = data.read()
        return content

    except FileNotFoundError as fi:
        print(fi)
        

    finally:
        try:
            data.close()
        except:
            print("Execution completed")

print(readfile('data.txt'))
```

### Assignment 3: Handling Multiple Exceptions

Write a function that takes a list of integers and returns their sum. Use try, except, and finally blocks to handle TypeError if a non-integer value is encountered and print an appropriate message.

```python
def sum_list (lst):
    sum = 0
    try:
        for i in lst:
            sum = sum + i

    except TypeError as ex:
        print(ex)
        sum = None

    finally:
        print("Execution is completed")
        


    return sum

# a = [1,2,3,4,5]
# print(sum_list(a))

# print(sum_list([1,2,3,4,5]))

print(sum_list([1, 2, 3, 'a']))  # None
```
