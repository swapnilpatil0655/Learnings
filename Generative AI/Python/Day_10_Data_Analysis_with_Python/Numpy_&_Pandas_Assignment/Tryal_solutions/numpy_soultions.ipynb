### Assignment 1: Array Creation and Manipulation

1. Create a NumPy array of shape (5, 5) filled with random integers between 1 and 20. Replace all the elements in the third column with 1.
2. Create a NumPy array of shape (4, 4) with values from 1 to 16. Replace the diagonal elements with 0.

```python
import numpy as np
```

```python
## 1. Create a NumPy array of shape (5, 5) filled with 
# random integers between 1 and 20. 

array = np.random.randint(1,21, size = (5,5))
print("Original array:")
print(array)

# Replace all the elements in the third column with 1.
array[:,2] = 1
print("Modified array:")
print(array)
```

```python
array  = np.random.randint(1,21, size=(5,5))
print(f"array: ")
print(array)

array[:,2] =1
print(f"Modified array:")
print(array)
```

```python
#2. Create a NumPy array of shape (4, 4) with values 
# from 1 to 16. 


array = np.arange(1,17).reshape((4,4))
print(array)

# Replace the diagonal elements with 0.
    # What is diagonal?
    # Positions affected:
    # (0,0), (1,1), (2,2), (3,3)

np.fill_diagonal(array,0)
print(array)
```

### Assignment 2: Array Indexing and Slicing

1. Create a NumPy array of shape (6, 6) with values from 1 to 36. Extract the sub-array consisting of the 3rd to 5th rows and 2nd to 4th columns.
2. Create a NumPy array of shape (5, 5) with random integers. Extract the elements on the border.

```python
#1. Create a NumPy array of shape (6, 6) with values from 1 to 36. 


array = np.arange(1,37).reshape((6,6))
print("Original array:")
print(array)

# Extract the sub-array consisting of the 
# 3rd to 5th rows and 2nd to 4th columns.

sub_array = array[2:5,1:4]
print("Sub-array:")
print(sub_array)
```

```python
## 2. Create a NumPy array of shape (5, 5) with random integers.
array = np.random.randint(1,10,size = (5,5))
print("array")
print(array)

# Extract the elements on the border.
# array[0:1,:] # first row
# array[-1:,:] # last row
# array[:,0:1] # first column
# array[:,-1:] # last column
border_elements = np.concatenate((array[0, :], array[-1, :], array[1:-1, 0], array[1:-1, -1]), axis=0)
print("Border elements:")
print(border_elements)

```
