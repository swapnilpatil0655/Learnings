# Data Visualization with Matplotlib

Matplotlib is a powerful and widely used data visualization library in Python. It provides a variety of tools for creating static, animated, and interactive visualizations. In this section, we will cover the basics of Matplotlib and how to create different types of plots.

```python
!pip install matplotlib
```

```python
import matplotlib.pyplot as plt
```

```python
x = [1,2,3,4,5,6]
y = [1,4,9,16,25,30]

## Create a line Plot
plt.plot(x,y)
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.title('Basic line Plot')
plt.show()
```

supported values are '-', '--', '-.', ':', 'None', ' ', '', 'solid', 'dashed', 'dashdot', 'dotted'

```python
x = [1,2,3,4,5,6]
y = [1,4,9,16,25,30]

## Create a customized line plot

plt.plot(x,y, color='red',linestyle='--', marker='o',linewidth=3,markersize=9)
```
