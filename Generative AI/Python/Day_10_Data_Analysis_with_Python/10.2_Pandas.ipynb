# Pandas_DataFrame and Series

Pandas is a powerful data manipulation library in Python, widely used for data analysis and data cleaning. It provides two primary data structure: Series and DataFrame. A Series is a one-dimensional array-like object, while a DataFrame is a two-dimensional, size-mutable, and potentially heterogeneous tabular data structure with labeled axes (row and columns).

### Series 
A Pandas Series is a one-dimensional array-like object that can hold any data type. It is similar to a column in a table.

```python
import pandas as pd
data = [1,2,3,4,5]
series = pd.Series(data)
print("Series \n", series)
```

```python
## Create a Series from dictionary
data = {'a':1, 'b': 2, 'c':3}
series_dict = pd.Series(data)
print("Series from dictionary \n", series_dict)

# a,b,c are index and 1,2,3 are values. We can also specify our own index.
```

```python
data = [10,20,30]
index = ['a','b','c']
pd.Series(data,index=index)
```

### DataFrame

`A DataFrame is a two-dimensional`,<br>`size-mutable, and potentially heterogeneous tabular data structure with labeled axes (row and columns)`. <br>`It is similar to a spreadsheet or SQL table.`

```python
## DataFrame
## Create a DataFrame from dictionary of list.

data={
    'Name': ['Swapnil', 'Rahul', 'Priya'],
    'Age': [25,30,22],
    'City': ['Pune', 'Mumbai', 'Delhi']
}
 # 3 columns and 3 rows. Name, Age, City are column names and Swapnil, Rahul, Priya are values in Name column. 25,30,22 are values in Age column and Pune, Mumbai, Delhi are values in City column.

df = pd.DataFrame(data)
print(df)
print(type(df))
```

```python
import numpy as np
np.array(df)
# We can also use numpy array to create a DataFrame. We need to specify column names and index.
# It will skip the first row and first column and will take the rest of the data to create a DataFrame. We can also specify our own index and column names. 
```

```python
## Create a DataFrame from list of dictionaries.
data = [
    {'Name': 'Swapnil', 'Age': 25, 'City': 'Pune'},
    {'Name': 'Rahul', 'Age': 30, 'City': 'Mumbai'},
    {'Name': 'Priya', 'Age': 22, 'City': 'Delhi'}
]

df = pd.DataFrame(data)
print(df)
print(type(df))
```
