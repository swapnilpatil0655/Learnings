# Data Visualization with Seaborn

Seaborn is a Python visualization library based on Matplotlib that provides a high-level interface for drawing attractive and informative statistical graphics. Seaborn helps in creating complex visualizations with just a few lines of code. In this lesson, we will cover the basics of Seaborn , Including creating various types of plots and customizing them.

````python
import seaborn as sns
````

Seaborn comes with several built-in datasets that can be used for practice and demonstration purposes. You can load these datasets using the `sns.load_dataset()` function. For example, to load the "tips" dataset, you can use the following code:

```python

https://github.com/mwaskom/seaborn-data

````python
### Basic Plotting with Seaborn
tips = sns.load_dataset('tips')
tips
````

## Create a scatter plot

````python
import matplotlib.pyplot as plt

sns.scatterplot(x='total_bill',y = 'tip',data = tips)
plt.title('Scatter Plot of Total Bill vs Tips')
plt.show()
````

## Line Plot

````python
sns.lineplot(x='size', y= 'total_bill',data = tips)
plt.title('Line Plot of Total Bill vs Size of the Party')
plt.show()
````

## Categorical Plot

Bar Plot

````python
sns.barplot(x='day', y='total_bill', data=tips)
plt.title('Bar Plot of Total Bill by Day of the Week')
plt.show()
````

## Box plot

````python
sns.boxplot(x='day', y='total_bill', data=tips)
````

## Violin Plot

````python
sns.violinplot(x='day', y='total_bill', data=tips)
````

## Histogram

````python
sns.histplot(tips['total_bill'],bins=20, kde=True)
# kde is for smoothing the histogram curve
````

````python
## KDE plot

sns.kdeplot(tips['total_bill'], fill=True)
````

## Pair Plot

pair plot is used to visualize the pairwise relationships between multiple variables in a dataset. It creates a grid of scatter plots for each pair of variables, along with histograms or KDE plots on the diagonal to show the distribution of each variable.

```python

````python
sns.pairplot(tips)
````

## Heatmap

````python
corr = tips[['total_bill','tip', 'size']].corr()
corr
````

````python
sns.heatmap(corr, annot=True, cmap='coolwarm')
````

````python
import pandas as pd
sales_df = pd.read_csv('sales_data.csv')
sales_df.head()
````

````python
## Plot total sales by product category
plt.figure(figsize=(10,6))
sns.barplot(x='Product Category', y='Total Revenue', data=sales_df, estimator=sum)
plt.title('Total Sales by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Total Revenue')
plt.show()
````

````python
## Plot total sales by product category
plt.figure(figsize=(10,6))
sns.barplot(x='Region', y='Total Revenue', data=sales_df, estimator=sum)
plt.title('Total Sales by Region')
plt.xlabel('Region')
plt.ylabel('Total Revenue')
plt.show()
````
