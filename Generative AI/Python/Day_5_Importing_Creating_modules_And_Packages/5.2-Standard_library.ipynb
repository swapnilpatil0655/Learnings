# Standard library overview

Python's Standard Library is a vast collection of modules and packages that come bundled with Python, providing a wide range of functionalities out of the box. Here's an overview of some of the most commonly used moduled and packages in Python Standard Library.

```python
import array
arr = array.array('i', [1, 2, 3, 4, 5]) # 'i' ensures that the only integer values are stored in the array.
print(arr)
```

### 📌 Common Type Codes in array

Here are the most important ones:

| Code  | Type           | Meaning                     |
| ----- | -------------- | --------------------------- |
| `'b'` | Signed char    | Small integer (-128 to 127) |
| `'B'` | Unsigned char  | 0 to 255                    |
| `'h'` | Signed short   | Small integer               |
| `'H'` | Unsigned short | Positive small integer      |
| `'i'` | Signed int     | Integer                     |
| `'I'` | Unsigned int   | Positive integer            |
| `'l'` | Signed long    | Large integer               |
| `'L'` | Unsigned long  | Positive large integer      |
| `'f'` | Float          | Decimal numbers             |
| `'d'` | Double         | High precision decimal      |

```python
## Data serialiazation and deserialization
import json
data = {'name': 'Alice', 'age': 30, 'city': 'New York'}

json_str=json.dumps(data)
print(json_str) # convert the Python dictionary to a JSON string
print(type(json_str)) # <class 'str'>

parsed_data = json.loads(json_str) # convert the JSON string back to a Python dictionary
print(parsed_data) # {'name': 'Alice', 'age': 30, 'city': 'New York'}
print(type(parsed_data)) # <class 'dict'>
```

```python
# csv

import csv

with open('example.csv', mode = 'w',newline='') as file: # open a new file named 'example.csv' in write mode.
    writer=csv.writer(file)                                     # create a csv writer object
    writer.writerow(['Name', 'Age', 'City'])                    # write the header row
    writer.writerow(['Alice', 30, 'New York'])                  # write a data row
    writer.writerow(['Bob', 25, 'Los Angeles'])                 # write another data row

with open('example.csv', mode='r') as file:             # open the file in read mode
    reader=csv.reader(file)                             # create a csv reader object
    for row in reader:                                  # iterate through each row in the csv file
        print(row)                                      # print the row as a list 
```

```python
## datetime

from datetime import datetime,timedelta 

now = datetime.now()
print(now)

yesterday=now-timedelta(days=1)
print(yesterday)
```

```python
#time
import time
print(time.time())

time.sleep(2)       # This code will sleep time for 2 sec
print(time.time())
```

```python
## Regular expression 

import re

pattern = r'\d+' # r= read where '\d+' digit pattern.
text='There are 123 apples 456'
match=re.search(pattern,text)
print(match.group())

# but this pattern code only identify the first digit continue value.
# that why 456 is not dected
```
