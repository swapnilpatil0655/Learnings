# Importing Modules in Python: Modules and Packages

In Python, modules and packages help organize and reuse code. Here's a comprehensive guide on how to import them.

```python
import math
math.sqrt(16)
```

```python
from math import sqrt,pi
print(sqrt(25))
print(sqrt(16))
print(pi)
```

```python
import numpy as np
np.array([1,2,3,4,5])
```

```python
from math import *
print(sqrt(16))
print(pi)
```

I have created the math.py file in package folder now I will import the math package and functions

```python
from package.maths import addition #package is the name of the folder and maths is the name of the file
addition(5,10)
```
