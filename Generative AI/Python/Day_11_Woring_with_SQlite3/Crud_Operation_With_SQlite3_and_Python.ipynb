# SQL and SQLite

SQL (Structured Query Language) is a standard language for managing and manipulating relational databases. SQLite is a self-contained, serverless, and zero-confifuration database engine that is widely used for embedded database systems. In this lesson, we will cover the basics of SQL and SQLite, including creating databases, tables, and performing various SQL operations.

````python
import sqlite3
````

````python
## Connect to an SQLite database (or create it if it doesn't exist)
connection = sqlite3.connect('example.db')
connection
````

````python
cursor = connection.cursor()
## cursor is used to execute SQL commands and queries
````

````python
## Create a Table
cursor.execute('''
Create Table If not exists employees(
               id integer primary key,
               name text not null,
               age integer,
               department text)
''')

connection.commit()  ## commit the changes to the database
````

````python
cursor.execute('select * from employees')
````

````python
## Insert Data into the Table
cursor.execute('''
Insert into employees(name,age,department)
        values('John Doe',30,'HR'),
      ('Jane Smith',25,'IT'),
      ('Alice Johnson',28,'Finance')
''')

## Insert Data into the Table
cursor.execute('''
Insert into employees(name,age,department)
            values('krishna',30,'HR')
''')

connection.commit()  ## commit the changes to the database
````

````python
## Query the data from the table
cursor.execute('select * from employees')
rows = cursor.fetchall()  ## fetch all rows from the result of the query

## print the queried data
for row in rows:
    print(row)
````

````python
## Update the data in the table
cursor.execute('''
update employees
               set age = 31
               where name = 'John Doe'
''')

connection.commit()  ## commit the changes to the database
````

````python
## Query the data from the table
cursor.execute('select * from employees')
rows = cursor.fetchall()  ## fetch all rows from the result of the query

## print the queried data
for row in rows:
    print(row)
````

````python
## Delete data from the table
cursor.execute('''
delete from employees
               where name = 'Jane Smith'
''')

connection.commit()  ## commit the changes to the database
````

````python
## Query the data from the table
cursor.execute('select * from employees')
rows = cursor.fetchall()  ## fetch all rows from the result of the query

## print the queried data
for row in rows:
    print(row)
````

````python
## Working with sales data
## Connect to an SQLite database
connection = sqlite3.connect('new_sales_data.db')
cursor = connection.cursor()

#Create a table for sales data
cursor.execute(''' 
create table if not exists sales(
               id integer primary key,
               date text not null,
                product text not null,
                quantity integer,
                region text)

''')

## Insert sample sales data
sales_data = [
    ('2024-01-01', 'Product A', 10, 'North'),
    ('2024-01-02', 'Product B', 20, 'South'),
    ('2024-01-03', 'Product C', 15, 'East'),
    ('2024-01-04', 'Product A', 5, 'West'),
    ('2024-01-05', 'Product B', 25, 'North')
]

cursor.executemany('''
insert into sales(date, product, quantity, region)
        values(?,?,?,?)
''', sales_data)

connection.commit()  ## commit the changes to the database
````

````python
cursor.execute('select * from sales')
rows = cursor.fetchall()  ## fetch all rows from the result of the query

## print the queried data
for row in rows:
    print(row)
````

````python
## close the database connection
connection.close()
````
