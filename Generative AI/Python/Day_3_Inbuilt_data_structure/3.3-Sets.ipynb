# Sets

Sets are a built-in data type in Python used to store collections of unique items. They are unordered, meaning that the elements do not follow a specific order, and they do not allow duplicate elements. Sets are useful for membership tests, eliminating duplicate entries, and performing mathematical set operations link union, intersection, difference, and symmetric difference. 

```python
##create a set
my_set = {1, 2, 3, 4, 5}
print(my_set)
print(type(my_set))
```

```python
my_empty_set = set()
print(type(my_empty_set))
```

```python
my_set = set([1, 2, 3, 4, 5,6])
print(type(my_set))
```

```python
my_set = set([1, 2, 3, 5,6,4, 5,6])
print(my_set)
```

```python
# Basic set operations
## Adding and removing elements
my_set.add(7)
print(my_set)

# Duplicate elements are not added to the set
my_set.add(7)
print(my_set)      # Output will still be {1, 2, 3, 4, 5, 6, 7}
```

```python
# Remove an element
my_set.remove(3)
print(my_set)
```

```python
my_set.remove(10)  # This will raise a KeyError since 10 is not in the set
print(my_set)
```

```python
my_set.discard(10)  # This will not raise an error, even if 10 is not in the set
print(my_set)
```

```python
## Pop method removes and returns an arbitrary element from the set. If the set is empty, it raises a KeyError.

removed_element = my_set.pop()
print(f"Removed element: {removed_element}")
print(my_set)
```

```python
## Clear all the elelments from the set
my_set.clear()
print(my_set)
```

```python
# my_set = set([1, 2, 3, 4, 5])   OR
my_set = {1, 2, 3, 4, 5}
print(type(my_set))
```

```python
## Set membership test
my_set = {1, 2, 3, 4, 5}
print(3 in my_set)  # Output: True
print(6 in my_set)  # Output: False
```

```python
# Mathematical Operation
set1 = {1, 2, 3, 4, 5,6}
set2 = {4, 5, 6, 7, 8,9}

### Union of sets
union_set = set1.union(set2)
print(f"Union of set1 and set2: {union_set}")


### Intersection of sets
intersection_set = set1.intersection(set2)
print(f"Intersection of set1 and set2: {intersection_set}")

set1.intersection_update(set2)
print(f"Set1 after intersection update with set2: {set1}")
```

```python
set1 = {1, 2, 3, 4, 5,6}
set2 = {4, 5, 6, 7, 8,9}

### Difference of sets
print(set1.difference(set2))
```

```python
set2.difference(set1) # This will return the elements that are in set2 but not in set1

set2.difference_update(set1) # This will update set2 to contain only the elements that are in set2 but not in set1
print(f"Set2 after difference update with set1: {set2}")

print(set2) 
```

```python
# Symmetric difference of sets
set1 = {1, 2, 3, 4, 5,6}
set2 = {4, 5, 6, 7, 8,9}
symmetric_difference = set1.symmetric_difference(set2) # This will return the elements that are in either set1 or set2 but not in both
print(f"Symmetric difference of set1 and set2: {symmetric_difference}")
```

### Set Method

```python
set1 = {1,2,3}
set2 = {4,5,6}

## is sybset
print(set1.issubset(set2)) 
"""his will return False since set1 is not a subset of set2. This means that not all elements of set1 are present in set2.
   Here, set2 is like a parent set and set1 is like a child set. 
   For set1 to be a subset of set2, all elements of set1 must be present in set2."""
```

```python
set1 = {1,2,3,4,5,6}
set2 = {4,5,6}
print(set1.issuperset(set2))
```

```python
lst = list(set(lst))  # Convert back to list if needed
print(lst)
```

```python
### Conting Unuque words in text
text = "This is a sample text with some sample words. This text is just for testing."
words = text.split() # Split the text into words    

## Convert the list of words to a set to get unique words
unique_words = set(words)
print(unique_words)
print(len(unique_words))
```
