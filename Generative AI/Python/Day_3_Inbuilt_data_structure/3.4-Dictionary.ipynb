### 1. Introdution to Dictionary
- Dictionaries are unordered collections of items. They stored data in key-value pairs. Keys must be unique and immutable (e.g., String, numbers, or tuples), which values can be of any type.

### 2. Creating Dictionay

```python
empty_dict = {}
print(type(empty_dict))  
```

```python
empty_dict = dict()
empty_dict
```

```python
student = {"name":"swapnil","age":30,"grade":"A"}
print(student)
print(type(student))
```

```python
# no duplicate keys allowed in dictionary it will override the previous value

student = {"name":"swapnil","age":31,"grade":"A"} 
print(student)
```

### 3. Accessing Dictonary Elements

```python
student = {"name":"swapnil","age":30,"grade":"A"}

print(student['grade'])  # Accessing value using key      OR 

print(student.get('name'))  # Accessing value using get method
print(student.get('last_name')) # Accessing non-existing key using get method returns None  \
print(student.get('Last_name', 'Not Available')) # Accessing non-existing key using get method with default value returns the default value
```

### 4. Modifying Dictionary Elements

- Dictionary are mutable, so you can add, update or delete elements.

```python
print(student)
```

```python
student["age"] = 33 # Updating existing key value pair
print(student)
student["country"] = "India" # Adding new key value pair to the dictionary if key is not already present in the dictionary
print(student)
```

```python
del student['grade'] # Deleting key value pair from the dictionary using del keyword
print(student)
```

### 5. Dictonary Methods

```python
keys = student.keys() # Get all keys from the dictionary
print(keys)
values = student.values() # Get all values from the dictionary
print(values)
```

```python
items = student.items() # Get all key value pairs from the dictionary as a list of tuples
print(items)
```

```python
# Shallow copy of dictionary
student_copy = student
print(student)
print(student_copy)
```

```python
student["name"]= "John" # Modifying the original dictionary also modifies the copied dictionary because both variables point to the same dictionary in memory
```

```python
student_copy = student
print(student)
print(student_copy)
```

```python
student_copy1 = student.copy() # Deep copy of dictionary using copy method
print(student)
print(student_copy1)
```

```python
student["name"] = "swapnil" # Modifying the original dictionary does not modify the copied dictionary because they are different dictionaries in memory
print(student)
print(student_copy1)
```

### 6. Iterating Over Dictionary 
- You can use loops to iterate over dictionary, key, values or items

```python
# Iterating over Keys

for key in student.keys():
    print(key)
```

```python
# Iterating over Values

for values in student.values():
    print(values)
```

```python
# iterating over key value pairs

for key, values in student.items():
    print(f"Key: {key}, Value: {values}")
```

### 7. Nested Dictionary

```python
student = {
    "student1" : {"name":"swapnil","age":30,"Country":"India"},
    "student2" : {"name":"nikhil","age":32,"Country":"India"}
}

print(student)
```

```python
print(student.student1.name) # Accessing nested dictionary value using dot notation is not possible it will raise an error
```

```python
print(student["student1"]["name"]) # Accessing nested dictionary value using keys
```

```python
# Iterating over nested dictionary

for student_id, student_info in student.items():
    print(f"{student_id}:{student_info}")
```

```python
for student_id, student_info in student.items():
    print(f"{student_id}:{student_info}")
    for key , values in student_info.items():
         print(f"{key}:{values}")
#    
```

### 8. Dictionary comphrehension

```python
square = {x:x**2 for x in range(5)}
print(square)
```

```python
## Conditional Dictionary Comprehension
even = {x:x**2 for x in range(10) if x%2==0}
print(even)
```

### 9. Practical Examples

```python
## Use a dictionary to count the frequency of elements in a list

numbers = [1,2,2,3,3,3,4,4,4,4]
frequency = {}
for numbers in numbers:
    if numbers in frequency:
        frequency[numbers] += 1
    else:
        frequency[numbers] = 1
print(frequency)
```

```python
numbers = [1,2,2,3,3,3,4,4,4,4]
frequency = {}

for num in numbers:
    if num in frequency:    
        frequency[num] += 1
    else:   
        frequency[num] = 1


print(frequency)
```

```python
# Merging two dictionaries in one   
dict1 = {"a":1,"b":2}
dict2 = {"c":3,"d":4}

merged_dict = {**dict1, **dict2} # Merging two dictionaries using unpacking operator
print(merged_dict)
```

```python
# Merge two dictionaries in one

dict1 = {"a":1,"b":2}
dict2 = {"b":3,"c":4}

merged_dict = {**dict1, **dict2} # Merging two dictionaries and updating the value of duplicate key with the value from the second dictionary
print(merged_dict)
```
