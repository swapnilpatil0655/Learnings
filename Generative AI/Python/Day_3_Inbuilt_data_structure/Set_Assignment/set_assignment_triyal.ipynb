### Assignment 3: Set Operations

Create two sets: one with the first 5 positive integers and another with the first 5 even integers. Perform and print the results of union, intersection, difference, and symmetric difference operations on these sets.

```python
set1 = set(range(1,6))
print(set1)
```

```python
set2 = set()
count = 0

for i in range(1,100):
    if count <  5:
        if i%2==0:
            set2.add(i)
            count = count + 1
print(set2)
            
```

```python
print(f"Set1 is first 5 positive integers: {set1}")
print(f"set2 is first 5 even integers: {set2}")
```

```python
# union

union = set1.union(set2)
print(f"union of set1 and set2 is: {union}")
```

```python
# Intersection

intersection = set1.intersection(set2)
print(f"intersection of set1 and set2 is: {intersection}")
```

```python
# Difference 

difference = set1.difference(set2)
print(f"difference of set1 and set2 is: {difference}") 
# it will give the element which are in set1 but not in set2
```

```python
# Symmetric

symmetirc = set1.symmetric_difference(set2)
print(f"Symmetric difference of set1 and set2 is: {symmetirc}")
# it will give the element which are in set1 or set2 but not in both
```

Create two sets: one with the first 5 positive integers and another with the first 3 positive integers. Check if the second set is a subset of the first set and if the first set is a superset of the second set. Print the results.

```python
set1 = (set(range(1,6)))
set2 = (set(range(1,4)))

set2.issubset(set1) # True because all elements of set2 are in set1
set1.issuperset(set2) # True because all elements of set2 are in set1
```
