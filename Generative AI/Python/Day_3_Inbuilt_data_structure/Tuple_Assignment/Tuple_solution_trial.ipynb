### Assignment 10: Tuple and List

Create a tuple with the first 5 positive integers. Convert it to a list, append the number 6, and convert it back to a tuple. Print the resulting tuple.

```python
tup = (1, 2, 3, 4, 5)
lst2 = list(tup)
lst2.append(6)
print(lst2)

tup = tuple(lst2)
print(tup)

```

```python
tup = (1, 2, 3, 4, 5)
lst = list(tup)
lst.append(6)
print(lst)

tup = tuple(lst)
print(tup)
```
